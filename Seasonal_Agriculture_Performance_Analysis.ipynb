```python
df = pd.read_csv("seasonal_agriculture_performance_dataset.csv")

print("Dataset loaded successfully!")
```

```python
df.head()
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
print(df.columns.tolist())
```

```python
df.isnull().sum()
```

```python
df['Rainfall_mm'] = df['Rainfall_mm'].fillna(
    df['Rainfall_mm'].median()
)

df['Soil_Moisture_pct'] = df['Soil_Moisture_pct'].fillna(
    df['Soil_Moisture_pct'].median()
)

df['Yield_Tonnes_Ha'] = df['Yield_Tonnes_Ha'].fillna(
    df['Yield_Tonnes_Ha'].median()
)

print("Missing values have been cleaned.")
```

```python
print(df.isnull().sum().sum())
```

```python
print(df['Season'].value_counts())
```

```python
season_yield = df.groupby('Season')['Yield_Tonnes_Ha'].mean()

print(season_yield)
```

```python
plt.figure(figsize=(8,5))

season_yield.plot(kind='bar')

plt.title('Average Agricultural Yield by Season')
plt.xlabel('Season')
plt.ylabel('Average Yield (Tonnes/Ha)')

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
season_profit = df.groupby('Season')['Profit_INR'].mean()

print(season_profit)
```

```python
plt.figure(figsize=(8,5))

season_profit.plot(kind='bar')

plt.title('Average Profit by Season')
plt.xlabel('Season')
plt.ylabel('Average Profit (INR)')

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
season_rainfall = df.groupby('Season')['Rainfall_mm'].mean()

print(season_rainfall)
```

```python
plt.figure(figsize=(8,5))

season_rainfall.plot(kind='bar')

plt.title('Average Rainfall by Season')
plt.xlabel('Season')
plt.ylabel('Rainfall (mm)')

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```
